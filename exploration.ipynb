```python
import fastf1
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

CACHE_DIR = "/Users/jamesburnard/Desktop/Projects/Pitwall-F1/cache"
fastf1.Cache.enable_cache(CACHE_DIR)

pd.set_option('display.max_columns', None)

year = 2024
gp = 'Silverstone'
# Define the sessions across the different days
days_sessions = ['FP1', 'FP2', 'FP3', 'Q', 'R'] 

# Dictionary to hold the data frames for each day
all_weekend_laps = {}

# 2. Loop through each session type
for session_type in days_sessions:
    print(f"Loading {gp} - {session_type}...")
    
    # Get and load the specific day's session
    session = fastf1.get_session(year, gp, session_type)
    session.load()
    
    # Store the lap data into our dictionary
    all_weekend_laps[session_type] = session.laps

```

```python
for driver in all_weekend_laps['R']['Driver'].unique():
    stint = all_weekend_laps['R'][all_weekend_laps['R']['Driver'] == driver]['Stint'].unique()
    for stint_item in stint:
        driver_stints = all_weekend_laps['R'][
            (all_weekend_laps['R']['Driver'] == driver) &
            (all_weekend_laps['R']['Stint'] == stint_item)
        ]

        lap_time = driver_stints['LapTime']
        compound = driver_stints['Compound'].unique()
        tyre_life = driver_stints['TyreLife'].unique()

        race_stints_df = all_weekend_laps['R'][
            ['Driver', 'Stint', 'LapNumber', 'LapTime', 'Compound', 'TyreLife']
        ].copy()

        race_stints_df = race_stints_df.sort_values(
            ['Driver', 'Stint', 'LapNumber']
        ).reset_index(drop=True)

        
        race_stints_df['LapTimeSeconds'] = race_stints_df['LapTime'].dt.total_seconds()
        race_stints_df[['LapTime', 'LapTimeSeconds']].head()
        race_stints_df

race_stints_df[race_stints_df['Driver'] == 'HAM'].head(53)
```

```python
#exploring relationship between TyreLife & LapTime

sns.scatterplot(x=race_stints_df[race_stints_df['Driver'] == 'HAM']['TyreLife'],
                y=race_stints_df[race_stints_df['Driver'] == 'HAM']['LapTimeSeconds'],
                hue=race_stints_df[race_stints_df['Driver'] == 'HAM']['Stint'])

plt.title('Relationship Between Tyre Life and LapTime (Hamilton)')
plt.show()
```
