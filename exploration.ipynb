```python
import fastf1
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

CACHE_DIR = "/Users/jamesburnard/Desktop/Projects/Pitwall-F1/cache"
fastf1.Cache.enable_cache(CACHE_DIR)

pd.set_option('display.max_columns', None)

year = 2024
gp = 'Silverstone'
# Define the sessions across the different days
days_sessions = ['FP1', 'FP2', 'FP3', 'Q', 'R'] 

# Dictionary to hold the data frames for each day
all_weekend_laps = {}

# 2. Loop through each session type
for session_type in days_sessions:
    print(f"Loading {gp} - {session_type}...")
    
    # Get and load the specific day's session
    session = fastf1.get_session(year, gp, session_type)
    session.load()
    
    # Store the lap data into our dictionary
    all_weekend_laps[session_type] = session.laps

```

```python
#Now I want to create a column called 'DidPit'
#Essentially telling us if the there was a pit in/out involved in the lap
#This will be useful for later since laps that contain pits have been identified as being sort unreliable.

all_weekend_laps['R']['DidPit'] = np.where(
    (all_weekend_laps['R']['PitInTime'].notna()) | (all_weekend_laps['R']['PitOutTime'].notna()),
    'Yes',
    'No'
)

#I only want to include laps where the track status is 1 (meaning the track was clear)
all_weekend_laps['R'] = all_weekend_laps['R'][all_weekend_laps['R']['TrackStatus'] == 1]

#General preprocessing, pulling necessary info

for driver in all_weekend_laps['R']['Driver'].unique():
    stint = all_weekend_laps['R'][all_weekend_laps['R']['Driver'] == driver][all_weekend_laps['R']['TrackStatus'] == 1]['Stint'].unique()
    for stint_item in stint:
        driver_stints = all_weekend_laps['R'][
            (all_weekend_laps['R']['Driver'] == driver) &
            (all_weekend_laps['R']['Stint'] == stint_item)
        ]

        lap_time = driver_stints['LapTime']
        compound = driver_stints['Compound'].unique()
        tyre_life = driver_stints['TyreLife'].unique()

        race_stints_df = all_weekend_laps['R'][
            ['Driver', 'Stint', 'LapNumber', 'LapTime', 'Compound', 'TyreLife', 'DidPit', 'TrackStatus']
        ].copy()

        race_stints_df = race_stints_df.sort_values(
            ['Driver', 'Stint', 'LapNumber']
        ).reset_index(drop=True)

        
        race_stints_df['LapTimeSeconds'] = race_stints_df['LapTime'].dt.total_seconds()
        race_stints_df[['LapTime', 'LapTimeSeconds']].head()
       

race_stints_df = race_stints_df.drop(race_stints_df[race_stints_df['DidPit'] == 'Yes'].index)
race_stints_df[race_stints_df['Driver'] == 'HAM']

```

```python
#exploring relationship between TyreLife & LapTime

sns.scatterplot(x=race_stints_df[race_stints_df['Driver'] == 'HAM']['TyreLife'],
                y=race_stints_df[race_stints_df['Driver'] == 'HAM']['LapTimeSeconds'],
                hue=race_stints_df[race_stints_df['Driver'] == 'HAM']['Stint'])

plt.title('Relationship Between Tyre Life and LapTime (Hamilton)')
plt.show()
```

```python
#Now I need to understand what Which laps actually represent normal racing pace, 
#and which laps should not be used to estimate tire degradation?
#Do do this (and with the understanding that F1 Lap data is not normally distributed)
#I will be using the IQR method to identify outliers in lap times.
#First I will create a boxplot to understand the distribution and potential outliers in lap times.
plt.figure(figsize=(10, 6))
sns.boxplot(x=race_stints_df['LapTimeSeconds'])
plt.title('Boxplot of Lap Times')
plt.xlabel('Lap Time (Seconds)')
plt.show()

q1 = race_stints_df['LapTimeSeconds'].quantile(0.25)
q3 = race_stints_df['LapTimeSeconds'].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 * 1.5 * iqr




```

```python
q1 = race_stints_df['LapTimeSeconds'].quantile(0.25)
q3 = race_stints_df['LapTimeSeconds'].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# Filter out the outliers based on the IQR method
filtered_race_stints_df = race_stints_df[(race_stints_df['LapTimeSeconds'] >= lower_bound) & (race_stints_df['LapTimeSeconds'] <= upper_bound)]

print(lower_bound, upper_bound, filtered_race_stints_df.shape[0])
```
