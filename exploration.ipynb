```python
import fastf1
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

CACHE_DIR = "/Users/jamesburnard/Desktop/Projects/Pitwall-F1/cache"
fastf1.Cache.enable_cache(CACHE_DIR)

pd.set_option('display.max_columns', None)

year = 2024
gp = 'Silverstone'
# Define the sessions across the different days
days_sessions = ['FP1', 'FP2', 'FP3', 'Q', 'R'] 

# Dictionary to hold the data frames for each day
all_weekend_laps = {}

# 2. Loop through each session type
for session_type in days_sessions:
    print(f"Loading {gp} - {session_type}...")
    
    # Get and load the specific day's session
    session = fastf1.get_session(year, gp, session_type)
    session.load()
    
    # Store the lap data into our dictionary
    all_weekend_laps[session_type] = session.laps

```

```python
#Now I want to create a column called 'DidPit'
#Essentially telling us if the there was a pit in/out involved in the lap
#This will be useful for later since laps that contain pits have been identified as being sort unreliable.

race_df = all_weekend_laps['R'].copy()

# Identify pit-in and pit-out laps
race_df['DidPit'] = np.where(
    race_df['PitInTime'].notna() | race_df['PitOutTime'].notna(),
    'Yes',
    'No'
)

# Select relevant columns
race_stints_df = race_df[
    ['Driver', 'Stint', 'LapNumber', 'LapTime',
     'Compound', 'TyreLife', 'DidPit', 'TrackStatus', 'LapStartTime']
].copy()

# Convert lap times to seconds
race_stints_df['LapTimeSeconds'] = race_stints_df['LapTime'].dt.total_seconds()
# Sort observations
race_stints_df = race_stints_df.sort_values(
    ['Driver', 'Stint', 'LapNumber']
).reset_index(drop=True)

# Create a separate cleaned dataset
clean_race_stints_df = race_stints_df[
    (race_stints_df['TrackStatus'].astype(str) == '1') &
    (race_stints_df['DidPit'] == 'No')
].copy()

# Display results
print("Original rows:", len(race_stints_df))
print("Cleaned rows:", len(clean_race_stints_df))

display(clean_race_stints_df.head(10))
```

```python
#exploring relationship between TyreLife & LapTime

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=clean_race_stints_df[clean_race_stints_df['Driver'] == 'HAM'],
    x='TyreLife',
    y='LapTimeSeconds',
    hue='Stint'
)

plt.title('Relationship Between Tyre Life and Lap Time (Hamilton)')
plt.show()
```

```python
#Now I need to understand what Which laps actually represent normal racing pace, 
#and which laps should not be used to estimate tire degradation?
#Do do this (and with the understanding that F1 Lap data is not normally distributed)
#I will be using the IQR method to identify outliers in lap times.
#First I will create a boxplot to understand the distribution and potential outliers in lap times.
plt.figure(figsize=(10, 6))
sns.boxplot(x=clean_race_stints_df['LapTimeSeconds'])
plt.title('Boxplot of Lap Times')
plt.xlabel('Lap Time (Seconds)')
plt.show()




```

```python
q1 = clean_race_stints_df['LapTimeSeconds'].quantile(0.25)
q3 = clean_race_stints_df['LapTimeSeconds'].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
filtered_race_stints_df = clean_race_stints_df[(clean_race_stints_df['LapTimeSeconds'] >= lower_bound) & (clean_race_stints_df['LapTimeSeconds'] <= upper_bound)]


plt.figure(figsize=(10, 6))
sns.boxplot(x=filtered_race_stints_df['LapTimeSeconds'])
plt.title('Boxplot of Lap Times')
plt.xlabel('Lap Time (Seconds)')
plt.show()
```

```python
sns.scatterplot(data=filtered_race_stints_df[filtered_race_stints_df['Driver'] == 'HAM'], x='TyreLife', y='LapTimeSeconds', hue='Stint')
plt.title('Relationship Between Tyre Life and LapTime (Hamilton)')
plt.show()
```

```python
weather_df = session.weather_data.copy()

filtered_race_stints_df = filtered_race_stints_df.sort_values('LapStartTime')
weather_df = weather_df.sort_values('Time')

stints_with_weather_df = pd.merge_asof(
    filtered_race_stints_df,
    weather_df,
    left_on='LapStartTime',
    right_on='Time',
    direction='nearest'
)
stints_with_weather_df.head(100)
```

```python
ham_df = stints_with_weather_df[stints_with_weather_df['Driver'] == 'HAM'].sort_values('LapNumber')

# Create a stacked 2-row plot sharing the exact same X-axis timeline
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# -------------------------------------------------------------
# PANEL 1: Race Pace & Tire Stints
# -------------------------------------------------------------
sns.scatterplot(
    data=ham_df, 
    x='LapNumber', 
    y='LapTimeSeconds', 
    hue='Stint', 
    palette='muted', 
    ax=ax1,
    s=70,
    edgecolor='w',
    linewidth=0.5
)
ax1.set_title('Lewis Hamilton: Race Pace vs. Changing Weather Timeline', fontsize=14, pad=15)
ax1.set_ylabel('Lap Time (Seconds)', fontsize=11)
ax1.legend(title='Stint', bbox_to_anchor=(1.08, 1), loc='upper left')
ax1.grid(True, linestyle='--', alpha=0.5)

# -------------------------------------------------------------
# PANEL 2: Weather Evolution (Track Temp, Air Temp, & Rainfall)
# -------------------------------------------------------------
# Line 1: Track Temperature (Left Y-Axis)
sns.lineplot(
    data=ham_df, 
    x='LapNumber', 
    y='TrackTemp', 
    color='#e74c3c', 
    linewidth=2.5, 
    label='Track Temp', 
    ax=ax2
)
ax2.set_ylabel('Track Temp (°C)', color='#e74c3c', fontsize=11)
ax2.tick_params(axis='y', labelcolor='#e74c3c')
ax2.grid(True, linestyle='--', alpha=0.5)

# Line 2: Air Temperature (Right Y-Axis via twinx)
ax2_twin = ax2.twinx()
sns.lineplot(
    data=ham_df, 
    x='LapNumber', 
    y='AirTemp', 
    color='#3498db', 
    linewidth=1.5, 
    linestyle='--',
    label='Air Temp', 
    ax=ax2_twin
)
ax2_twin.set_ylabel('Air Temp (°C)', color='#3498db', fontsize=11)
ax2_twin.tick_params(axis='y', labelcolor='#3498db')

# Cool Highlight: Shading the background area where Rainfall > 0
# This fills a light blue bar over the rain laps so you instantly spot track moisture
ax2.fill_between(
    ham_df['LapNumber'], 
    ham_df['TrackTemp'].min()-2,             # 2. The lower Y-limit of the shading
    ham_df['TrackTemp'].max()+2, 
    where=ham_df['Rainfall'], 
    color='#3498db', 
    alpha=0.15, 
    label='Rainfall Active'
)

# Align legends for the weather plots cleanly
ax2.legend(loc='upper left')
ax2_twin.legend(loc='upper right')
ax2.set_xlabel('Lap Number', fontsize=11)

# Ensure tight layout formatting so axes labels aren't clipped
plt.tight_layout()
plt.show()
```
